```python
import random

NUM_PARTICLES = 5
MAX_ITERATIONS = 10
MIN_GREEN_TIME = 10.0
MAX_GREEN_TIME = 60.0

W = 0.5
C1 = 1.5
C2 = 1.5

def run_traffic_simulation(position):
    optimal_time = 35.0
    return sum((t - optimal_time) ** 2 for t in position)

DIMENSIONS = 4

positions = [[random.uniform(MIN_GREEN_TIME, MAX_GREEN_TIME) for _ in range(DIMENSIONS)] for _ in range(NUM_PARTICLES)]
velocities = [[random.uniform(-1, 1) for _ in range(DIMENSIONS)] for _ in range(NUM_PARTICLES)]

pbest_positions = list(positions)
pbest_delays = []

gbest_position = None
gbest_delay = float('inf')

for i in range(NUM_PARTICLES):
    delay = run_traffic_simulation(positions[i])
    pbest_delays.append(delay)

    if delay < gbest_delay:
        gbest_delay = delay
        gbest_position = list(positions[i])

for iteration in range(MAX_ITERATIONS):
    for i in range(NUM_PARTICLES):
        total_vehicle_delay = run_traffic_simulation(positions[i])

        if total_vehicle_delay < pbest_delays[i]:
            pbest_delays[i] = total_vehicle_delay
            pbest_positions[i] = list(positions[i])

        if total_vehicle_delay < gbest_delay:
            gbest_delay = total_vehicle_delay
            gbest_position = list(positions[i])

    for i in range(NUM_PARTICLES):
        for d in range(DIMENSIONS):
            r1 = random.random()
            r2 = random.random()

            velocities[i][d] = (W * velocities[i][d]) + \
                               (C1 * r1 * (pbest_positions[i][d] - positions[i][d])) + \
                               (C2 * r2 * (gbest_position[d] - positions[i][d]))

            positions[i][d] = positions[i][d] + velocities[i][d]

            if positions[i][d] < MIN_GREEN_TIME:
                positions[i][d] = MIN_GREEN_TIME
            elif positions[i][d] > MAX_GREEN_TIME:
                positions[i][d] = MAX_GREEN_TIME

    print(f"Iteration {iteration + 1}/{MAX_ITERATIONS} - Best Delay: {gbest_delay:.2f}")

print("\n=== Optimization Complete ===")
print(f"Optimal Signal Timings (gbest): {[round(t, 2) for t in gbest_position]} seconds")
print(f"Lowest Total Vehicle Delay: {gbest_delay:.2f}")
```
