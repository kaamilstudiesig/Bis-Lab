```python
import random

POPULATION_SIZE = 50
GENERATIONS = 100
MUTATION_RATE = 0.1
INITIAL_SURVIVAL_THRESHOLD = 45.0
TRAFFIC_DEMAND = [3.2, 4.5, 1.8, 2.9]

class SignalSolution:
    def __init__(self, green_times=None):
        if green_times is None:
            self.green_times = [random.randint(10, 90) for _ in range(4)]
        else:
            self.green_times = green_times

        self.total_cycle = 120
        self.red_times = [self.total_cycle - g for g in self.green_times]
        self.fitness = 0.0
        self.avg_waiting_time = 0.0

    def calculate_fitness(self):
        total_waiting_time = 0
        for g, r, demand in zip(self.green_times, self.red_times, TRAFFIC_DEMAND):
            delay = (demand * (r ** 2)) / (2 * self.total_cycle) + max(0, (demand - g) * 2)
            total_waiting_time += delay

        self.avg_waiting_time = total_waiting_time / len(TRAFFIC_DEMAND)
        self.fitness = 1000 / (self.avg_waiting_time + 1)

def breed(parent1, parent2):
    crossover_point = random.randint(1, 3)
    child1_genes = parent1.green_times[:crossover_point] + parent2.green_times[crossover_point:]
    child2_genes = parent2.green_times[:crossover_point] + parent1.green_times[crossover_point:]
    return SignalSolution(child1_genes), SignalSolution(child2_genes)

def mutate(solution):
    for i in range(len(solution.green_times)):
        if random.random() < MUTATION_RATE:
            solution.green_times[i] = max(10, min(90, solution.green_times[i] + random.choice([-5, 5])))
    self_total_cycle = solution.total_cycle
    solution.red_times = [self_total_cycle - g for g in solution.green_times]

population = [SignalSolution() for _ in range(POPULATION_SIZE)]
current_threshold = INITIAL_SURVIVAL_THRESHOLD

print(f"Starting Evolution with Population Size: {len(population)}\n")

for generation in range(GENERATIONS):
    for solution in population:
        solution.calculate_fitness()

    survivors = [ind for ind in population if ind.avg_waiting_time <= current_threshold]

    if not survivors:
        population.sort(key=lambda x: x.avg_waiting_time)
        best_overall = population[0]
        current_threshold = best_overall.avg_waiting_time
        survivors = [ind for ind in population if ind.avg_waiting_time <= current_threshold]

    population = survivors
    population.sort(key=lambda x: x.fitness, reverse=True)
    best_solution = population[0]

    print(f"Gen {generation+1} | Threshold: {current_threshold:.2f}s | Surviving Pool: {len(population)} | Best Wait: {best_solution.avg_waiting_time:.2f}s")

    if best_solution.avg_waiting_time < 15.0:
        print("Optimal solution threshold achieved!")
        break

    next_generation = []
    if len(population) > 1:
        next_generation.extend(population[:2])
    else:
        next_generation.extend([population[0], population[0]])

    while len(next_generation) < POPULATION_SIZE:
        if len(population) > 1:
            p1, p2 = random.sample(population[:15], 2) if len(population) > 5 else random.sample(population, 2)
        else:
            p1, p2 = population[0], population[0]

        c1, c2 = breed(p1, p2)
        mutate(c1)
        mutate(c2)
        next_generation.extend([c1, c2])

    population = next_generation[:POPULATION_SIZE]

if population:
    population[0].calculate_fitness()
    print("\n=== OPTIMAL SIGNAL TIMING FOUND ===")
    print(f"Optimized Green Times (seconds): {population[0].green_times}")
    print(f"Derived Red Times (seconds):     {population[0].red_times}")
    print(f"Resulting Avg Waiting Time:      {population[0].avg_waiting_time:.2f} seconds")
```
